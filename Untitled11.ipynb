```python
import pandas as pd

from sklearn.linear_model import LinearRegression
df = pd.DataFrame({
    "length": [5, 6, 7, 8, 9, 10],

    "alanine_count": [1, 1, 2, 2, 3, 3],

    "target": [5.1, 5.6, 6.0, 6.4, 7.0, 7.3]
})
X = df[["length", "alanine_count"]]

y = df["target"]
model = LinearRegression()
model.fit(X, y)
print("Coefficients:", model.coef_)

print("Intercept:", model.intercept_)
```
